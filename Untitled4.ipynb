````python
import os, json, uuid, time, shutil, asyncio, subprocess, re, threading
from datetime import datetime
from pathlib import Path

import gradio as gr
from groq import Groq
from pydub import AudioSegment
from faster_whisper import WhisperModel

DATA_DIR = Path("/content/ai_voice_sales_bot_data")
DATA_DIR.mkdir(parents=True, exist_ok=True)
HISTORY_FILE = DATA_DIR / "call_history.json"
CHAT_MODEL = "llama-3.3-70b-versatile"
PRODUCT_INFO = """TaskFlow CRM is a fictional SaaS product for small businesses. It organizes customer leads, tracks follow-ups, keeps team notes in one place, and provides a simple sales pipeline dashboard. The demo offer is a 14-day trial. Do not invent prices or promise features beyond these details."""
VALID_STATUSES = ["Converted", "Interested", "Follow-up Required", "Not Interested", "Demo Requested", "No Response"]

_groq = None
if os.environ.get("GROQ_API_KEY"):
    try:
        _groq = Groq(api_key=os.environ["GROQ_API_KEY"])
    except Exception as e:
        print("Groq client setup warning:", str(e)[:240])

_whisper = None
_whisper_lock = threading.Lock()
def get_whisper():
    global _whisper
    with _whisper_lock:
        if _whisper is None:
            print("Loading local Whisper model (first run may take a few minutes)...")
            _whisper = WhisperModel("small", device="cpu", compute_type="int8")
    return _whisper

def load_history():
    try:
        if HISTORY_FILE.exists():
            return json.loads(HISTORY_FILE.read_text(encoding="utf-8"))
    except Exception as e:
        print("History read warning:", str(e)[:180])
    return []

def save_history(items):
    try:
        HISTORY_FILE.write_text(json.dumps(items, ensure_ascii=False, indent=2), encoding="utf-8")
    except Exception as e:
        print("History save warning:", str(e)[:180])

def format_duration(seconds):
    seconds = max(0, int(seconds))
    return f"{seconds//60:02d}:{seconds%60:02d}"

def transcribe_audio(audio_path):
    if not audio_path:
        return ""
    try:
        model = get_whisper()
        segments, _ = model.transcribe(audio_path, beam_size=1, vad_filter=True)
        text = " ".join(s.text.strip() for s in segments).strip()
        return text
    except Exception as e:
        raise RuntimeError("Local transcription failed. Try a shorter, clearer clip or use the text input. Details: " + str(e)[:220])

def browser_tts_text(text):
    # The hidden output is read by a small browser-side JS callback after each response.
    return (text or "")[:1200]

def make_ai_audio(text, out_path):
    """Generate an audio segment for the saved call using free edge-tts if available."""
    try:
        import edge_tts
        async def _save():
            communicate = edge_tts.Communicate(text, voice="en-US-JennyNeural", rate="+0%")
            await communicate.save(str(out_path))
        asyncio.run(_save())
        return str(out_path) if Path(out_path).exists() and Path(out_path).stat().st_size > 0 else None
    except Exception as e:
        print("AI audio recording segment unavailable; transcript is still saved:", str(e)[:160])
        return None

def fallback_reply(user_text, consent):
    low = user_text.lower()
    if consent is None:
        if any(x in low for x in ["don't", "do not", "no", "not comfortable", "decline"]):
            return "No problem. I will not record this conversation. We can continue without saving audio. What would you like to know about TaskFlow CRM?"
        if any(x in low for x in ["yes", "yeah", "sure", "okay", "ok", "agree", "go ahead", "fine"]):
            return "Thank you for confirming. TaskFlow CRM helps small businesses organize leads, follow-ups, team notes, and their sales pipeline. What is the biggest challenge in your current sales process?"
        return "Before we continue, may I have your permission to record this demo conversation? Please say yes or no, or type your answer."
    if any(x in low for x in ["demo", "show me", "demonstration"]):
        return "Absolutely, I can mark that you would like a demo. What day or time usually works best for you?"
    if any(x in low for x in ["price", "cost", "expensive", "budget"]):
        return "I understand that budget matters. This demo offer includes a 14-day trial, and I don't want to invent a price. Would trying it first help you evaluate whether it fits your needs?"
    if any(x in low for x in ["not interested", "no thanks", "not needed"]):
        return "Thanks for being direct. I will mark this as not interested. Is there anything else you would like me to note?"
    if any(x in low for x in ["yes", "interested", "sounds good", "trial", "sign up", "buy"]):
        return "Great! TaskFlow CRM has a 14-day trial. Would you like to request a demo or discuss trying the product?"
    return "Thanks for sharing that. TaskFlow CRM helps organize leads, follow-ups, team notes, and a sales pipeline dashboard. What would you most like to improve in your current workflow?"

def ai_reply(messages, consent):
    if not messages:
        return "Hi! I'm the TaskFlow CRM sales assistant. Before we begin, may I have your permission to record this demo conversation? You can say yes or no."
    if _groq is None:
        return fallback_reply(messages[-1]["content"], consent)
    try:
        api_messages = [{"role":"system", "content": (
            "You are a warm, concise SaaS sales assistant for TaskFlow CRM. Ask one question at a time. "
            "Never invent prices or product features. The product details are: " + PRODUCT_INFO + " "
            "The customer has " + ("not consented to recording; do not imply audio is being saved." if consent is False else "consented to recording." if consent is True else "not yet answered the recording-consent question; ask only for consent before selling." )
            + " Keep each response under 90 words. If consent is not yet given, ask for clear yes/no consent."
        )}]
        for m in messages[-12:]:
            api_messages.append({"role": m["role"], "content": m["content"]})
        result = _groq.chat.completions.create(model=CHAT_MODEL, messages=api_messages, temperature=0.4, max_tokens=180)
        content = result.choices[0].message.content
        return content.strip() if content else fallback_reply(messages[-1]["content"], consent)
    except Exception as e:
        print("Groq response fallback:", str(e)[:220])
        return fallback_reply(messages[-1]["content"], consent)

def update_consent_from_text(state, text):
    if state.get("recording_consent") is not None:
        return state.get("recording_consent")
    low = text.lower().strip()
    no_words = ["no", "nope", "don't", "do not", "not comfortable", "not agree", "decline"]
    yes_words = ["yes", "yeah", "yep", "sure", "okay", "ok", "i agree", "agree", "go ahead", "that's fine", "fine"]
    # Detect negative intent before positive substrings.
    if any(x in low for x in no_words):
        state["recording_consent"] = False
    elif any(x in low for x in yes_words):
        state["recording_consent"] = True
    return state.get("recording_consent")

def add_ai_audio_to_state(state, text):
    if state.get("recording_consent") is True:
        out = DATA_DIR / f"{state['call_id']}_ai_{len(state['audio_segments']):03d}.mp3"
        made = make_ai_audio(text, out)
        if made:
            state["audio_segments"].append(made)

def start_call():
    call_id = datetime.now().strftime("%Y%m%d_%H%M%S_") + uuid.uuid4().hex[:6]
    greeting = "Hi! I'm the TaskFlow CRM sales assistant. Before we begin, may I have your permission to record this demo conversation? Please say yes or no."
    state = {
        "call_id": call_id, "started_at": datetime.now().isoformat(timespec="seconds"),
        "start_epoch": time.time(), "messages": [{"role":"assistant", "content": greeting}],
        "audio_segments": [], "recording_consent": None, "ended": False, "status": "In Progress",
        "transcript": "", "recording_path": None
    }
    return state, state["messages"], "🟢 In progress — awaiting recording consent", "00:00", "Call started. The bot is asking for recording consent.", browser_tts_text(greeting), "The summary appears after you end the call."

def get_customer_text(audio_path, typed_text):
    if typed_text and typed_text.strip():
        return typed_text.strip(), audio_path
    if audio_path:
        return transcribe_audio(audio_path), audio_path
    return "", None

def process_turn(audio_path, typed_text, state):
    if not state or state.get("ended"):
        return state, (state or {}).get("messages", []), "🔴 No active call. Press Start Call.", "00:00", "Start a call before sending a message.", "", "The summary appears after you end the call."
    try:
        user_text, audio_path = get_customer_text(audio_path, typed_text)
    except Exception as e:
        return state, state["messages"], "🟢 In progress", format_duration(time.time()-state["start_epoch"]), str(e), "", "The summary appears after you end the call."
    if not user_text:
        return state, state["messages"], "🟢 In progress", format_duration(time.time()-state["start_epoch"]), "No speech detected. Try speaking clearly or use the text input.", "", "The summary appears after you end the call."

    old_consent = state.get("recording_consent")
    new_consent = update_consent_from_text(state, user_text)
    # Only save audio after consent was already given before this turn. The consent answer itself is not recorded.
    if old_consent is True and audio_path:
        try:
            saved = DATA_DIR / f"{state['call_id']}_customer_{len(state['audio_segments']):03d}.wav"
            AudioSegment.from_file(audio_path).export(saved, format="wav")
            state["audio_segments"].append(str(saved))
        except Exception as e:
            print("Customer audio clip could not be added to recording:", str(e)[:160])

    state["messages"].append({"role":"user", "content": user_text})
    reply = ai_reply(state["messages"], new_consent)
    state["messages"].append({"role":"assistant", "content": reply})
    if state.get("recording_consent") is True:
        add_ai_audio_to_state(state, reply)
    status = "🟢 In progress" + (" — recording consent granted" if state.get("recording_consent") is True else " — not recording audio" if state.get("recording_consent") is False else " — awaiting consent")
    activity_text = "Transcript updated." if _groq else "Transcript updated using fallback sales logic (Groq key not configured)."
    if state.get("recording_consent") is False:
        activity_text += " Recording is disabled because consent was not granted."
    return state, state["messages"], status, format_duration(time.time()-state["start_epoch"]), activity_text, browser_tts_text(reply), "The summary appears after you end the call."

def analyse_call(state):
    msgs = state.get("messages", [])
    transcript = "\n".join(("Customer: " if m["role"] == "user" else "AI: ") + m["content"] for m in msgs)
    if _groq and any(m["role"] == "user" for m in msgs):
        try:
            prompt = "Analyze this sales call. Return ONLY valid JSON with keys summary (string), customer_requirements (array of strings), objections (array of strings), next_steps (array of strings), sales_status (one of Converted, Interested, Follow-up Required, Not Interested, Demo Requested, No Response). Use evidence from the transcript only. Transcript:\n" + transcript
            res = _groq.chat.completions.create(model=CHAT_MODEL, messages=[{"role":"system","content":"You are a careful sales operations analyst. Return valid JSON only."},{"role":"user","content":prompt}], temperature=0.1, max_tokens=350)
            raw = res.choices[0].message.content.strip()
            raw = re.sub(r"^```(?:json)?\s*|\s*```$", "", raw, flags=re.I)
            data = json.loads(raw)
            if data.get("sales_status") not in VALID_STATUSES:
                data["sales_status"] = "Follow-up Required" if any(m["role"] == "user" for m in msgs) else "No Response"
            for k in ["customer_requirements", "objections", "next_steps"]:
                if not isinstance(data.get(k), list): data[k] = []
            if not isinstance(data.get("summary"), str): data["summary"] = "Call ended; see transcript."
            return data
        except Exception as e:
            print("AI summary fallback:", str(e)[:220])
    user_text = " ".join(m["content"].lower() for m in msgs if m["role"] == "user")
    if not any(m["role"] == "user" for m in msgs):
        status = "No Response"
    elif any(x in user_text for x in ["not interested", "no thanks", "not needed"]):
        status = "Not Interested"
    elif any(x in user_text for x in ["demo", "demonstration", "show me"]):
        status = "Demo Requested"
    elif any(x in user_text for x in ["converted", "sign me up", "i will buy", "let's buy"]):
        status = "Converted"
    elif any(x in user_text for x in ["interested", "trial", "sounds good", "yes, interested"]):
        status = "Interested"
    else:
        status = "Follow-up Required"
    return {"summary":"Fallback summary generated from the transcript. Review the conversation for details.",
            "customer_requirements":[], "objections":[], "next_steps":["Review transcript and follow up manually."], "sales_status":status}

def finish_recording(state):
    if state.get("recording_consent") is not True or not state.get("audio_segments"):
        return None
    try:
        combined = AudioSegment.empty()
        for path in state["audio_segments"]:
            if path and Path(path).exists():
                combined += AudioSegment.from_file(path) + AudioSegment.silent(duration=250)
        if len(combined) == 0:
            return None
        out = DATA_DIR / f"{state['call_id']}_recording.mp3"
        combined.export(out, format="mp3", bitrate="96k")
        return str(out)
    except Exception as e:
        print("Recording assembly failed:", str(e)[:220])
        return None

def end_call(state):
    if not state:
        return state, [], "No call to end.", "00:00", "Press Start Call to begin.", None, "No call summary available.", refresh_history()
    if state.get("ended"):
        return state, state["messages"], "Call already ended.", format_duration(time.time()-state["start_epoch"]), "This call is already closed.", state.get("recording_path"), "Call already ended.", refresh_history()
    duration = format_duration(time.time()-state["start_epoch"])
    analysis = analyse_call(state)
    recording_path = finish_recording(state)
    transcript = "\n".join(("Customer: " if m["role"] == "user" else "AI: ") + m["content"] for m in state["messages"])
    record = {"call_id":state["call_id"], "started_at":state["started_at"], "duration":duration,
              "transcript":transcript, "messages":state["messages"], "recording_path":recording_path,
              "recording_consent":state.get("recording_consent"), **analysis}
    history = load_history(); history.insert(0, record); save_history(history)
    state["ended"] = True; state["status"] = analysis["sales_status"]; state["recording_path"] = recording_path; state["duration"] = duration
    summary = (f"### Summary\n{analysis.get('summary','')}\n\n"
               f"**Customer requirements:** {', '.join(analysis.get('customer_requirements', [])) or 'None identified'}\n\n"
               f"**Objections:** {', '.join(analysis.get('objections', [])) or 'None identified'}\n\n"
               f"**Next steps:** {', '.join(analysis.get('next_steps', [])) or 'None specified'}\n\n"
               f"**Sales status:** **{analysis.get('sales_status','Follow-up Required')}**\n\n"
               f"**Duration:** {duration}\n\n"
               f"**Recording:** {'Saved with consent.' if recording_path else 'No audio recording saved (consent not granted or audio generation unavailable). Transcript is saved.'}")
    status = f"🔴 Call ended — {analysis.get('sales_status','Follow-up Required')}"
    return state, state["messages"], status, duration, "Call saved to history.", recording_path, summary, refresh_history()

def refresh_history():
    history = load_history()
    if not history: return "No completed calls yet."
    lines=[]
    for item in history[:15]:
        lines.append(f"**{item.get('started_at','')}** · `{item.get('call_id','')}` · **{item.get('sales_status','')}** · {item.get('duration','')}\n\n{item.get('summary','')}\n\nTranscript:\n```text\n{item.get('transcript','')[:1600]}\n```\n\nRecording: {'saved' if item.get('recording_path') else 'not available'}")
    return "\n\n---\n\n".join(lines)

def tick_duration(state):
    if not state or not state.get("start_epoch"):
        return "00:00"
    if state.get("ended"):
        return state.get("duration", "00:00")
    return format_duration(time.time()-state["start_epoch"])
````

````python
BROWSER_SPEECH_JS = r"""(text) => {
  try {
    if (typeof text === 'string' && text.trim() && 'speechSynthesis' in window) {
      window.speechSynthesis.cancel();
      const utterance = new SpeechSynthesisUtterance(text);
      utterance.lang = 'en-US';
      utterance.rate = 1.0;
      window.speechSynthesis.speak(utterance);
    }
  } catch (e) { console.warn('Browser speech unavailable; use the chat transcript.', e); }
  return [];
}"""

with gr.Blocks(title="AI Voice Sales Bot") as demo:
    gr.Markdown("# 🎙️ AI Voice Sales Bot\nTalk to an AI sales assistant for **TaskFlow CRM**. The bot asks for recording consent before saving audio.")
    state = gr.State(None)
    with gr.Row():
        start_btn = gr.Button("▶ Start Call", variant="primary")
        end_btn = gr.Button("■ End Call", variant="stop")
        status_box = gr.Textbox(label="Call Status", value="Not started", interactive=False)
        timer_box = gr.Textbox(label="Call Duration", value="00:00", interactive=False)
    chat = gr.Chatbot(label="Live Conversation", height=360)
    with gr.Row():
        customer_audio = gr.Audio(label="Speak (record a short message, then submit)", sources=["microphone"], type="filepath")
        text_fallback = gr.Textbox(label="Text fallback", placeholder="Type what the customer says…", lines=2)
    with gr.Row():
        send_btn = gr.Button("Send Voice/Text Message", variant="primary")
        clear_input_btn = gr.Button("Clear Inputs")
    activity = gr.Markdown("Press **Start Call** to begin.")
    speech_text = gr.Textbox(label="Browser speech output", visible=False)
    gr.Markdown("### Call Summary")
    summary_box = gr.Markdown("The summary appears after you end the call.")
    recording_player = gr.Audio(label="Saved call recording (playback)", type="filepath", interactive=False)
    with gr.Accordion("Call History", open=False):
        history_box = gr.Markdown(refresh_history())
        refresh_btn = gr.Button("Refresh History")
    gr.Markdown("**Privacy note:** Do not share real customer information without appropriate consent. Recordings and history are stored in the current Colab runtime and may disappear when it resets.")

    start_event = start_btn.click(
        fn=start_call, inputs=[], outputs=[state, chat, status_box, timer_box, activity, speech_text, summary_box]
    )
    start_event.then(fn=None, inputs=[speech_text], outputs=[], js=BROWSER_SPEECH_JS)

    turn_event = send_btn.click(
        fn=process_turn,
        inputs=[customer_audio, text_fallback, state],
        outputs=[state, chat, status_box, timer_box, activity, speech_text, summary_box]
    )
    turn_event.then(fn=None, inputs=[speech_text], outputs=[], js=BROWSER_SPEECH_JS)
    text_fallback.submit(
        fn=process_turn,
        inputs=[customer_audio, text_fallback, state],
        outputs=[state, chat, status_box, timer_box, activity, speech_text, summary_box]
    ).then(fn=None, inputs=[speech_text], outputs=[], js=BROWSER_SPEECH_JS)

    end_btn.click(
        fn=end_call,
        inputs=[state],
        outputs=[state, chat, status_box, timer_box, activity, recording_player, summary_box, history_box]
    )
    clear_input_btn.click(fn=lambda: (None, ""), inputs=[], outputs=[customer_audio, text_fallback])
    refresh_btn.click(fn=refresh_history, inputs=[], outputs=[history_box])
    gr.Timer(1).tick(fn=tick_duration, inputs=[state], outputs=[timer_box], queue=False)

demo.launch(share=True, debug=False, theme=gr.themes.Soft())
````
